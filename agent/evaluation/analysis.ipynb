```python
import json
import pandas as pd
```

```python
data_list = []
with open(r'result_0417\evaluation.json', 'r', encoding='utf-8') as f:
    for line in f:
        data = json.loads(line.strip())
        data_list.append(data)
  
```

```python
df = pd.DataFrame(data_list)
```

```python
df.to_csv(r'result_0417\result_VRAG_vs_KP.csv', index=False, encoding='utf-8')
```

```python
# answer_1 = baseline
# answer_2 = ours



counter = 0

for i in df['Evaluation']:
  print(i)
  if i['Overall']['Winner'] == 'Answer 2':
    counter += 1

counter
```

```python
# Extract the required metrics and overall result from the Evaluation column
df_evaluation = df['Evaluation'].apply(lambda eval_dict: {
  'Completeness': eval_dict['Completeness']['Winner'],
  'Knowledge Alignment': eval_dict['Knowledge Alignment']['Winner'],
  'Diversity': eval_dict['Diversity']['Winner'],
  'Overall': eval_dict['Overall']['Winner']
})

# Convert the extracted data into a new dataframe
df_evaluation = pd.DataFrame(df_evaluation.tolist())

# Display the resulting dataframe
print(df_evaluation)

df_evaluation.to_csv(r'result_0417\result_VRAG_vs_KP_evaluation.csv', index=False, encoding='utf-8')
```
