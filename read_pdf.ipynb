```python
from langchain_unstructured import UnstructuredLoader

loader = UnstructuredLoader(
    file_path=pdf_file,
    strategy="hi_res",
    partition_via_api=True,
    coordinates=True,
)
docs = []
for doc in loader.lazy_load():
    docs.append(doc)
```

```python
print(len(docs))
```

```python
first_page_docs = [doc for doc in docs if doc.metadata.get("page_number") == 1]

for doc in first_page_docs:
    print(doc.page_content)
```

```python
unique_page_number = set()

for doc in docs:
    unique_page_number.add(doc.metadata.get("page_number"))

print(unique_page_number)
```

```python
unique_page_number = sorted(list(unique_page_number))
pages = []
final_documents = []
for page_num in unique_page_number:
    final_documents.append([doc for doc in docs if doc.metadata.get("page_number") == page_num])

print(final_documents)
```

```python
for i in final_documents:
    for page_content in i:
        print(page_content.page_content)
    print("---end of page---")
    print('\n\n\n')
```
